## 3. Variables

A variable is a name that refers to an object/value.

Python is **dynamically typed**: you do not normally declare a variable's type before assigning a value.

```python
age = 22
name = "Hemil"
```

### Reassignment

A variable can refer to objects of different types:

```python
x = 10
x = "Python"
```

This is valid, although changing the meaning/type of a variable unnecessarily can reduce readability.

### Naming rules

Valid:

```python
student_name = "Hemil"
_age = 22
name2 = "ABC"
```

Invalid:

```python
2name = "ABC"       # cannot start with a number
student-name = "A" # '-' is subtraction, not part of a variable name
```

### Naming convention

Use `snake_case` for variables and functions:

```python
first_name = "Hemil"
total_marks = 85
```

Constants are conventionally written in uppercase:

```python
MAX_RETRIES = 3
PI = 3.14159
```

Python does not enforce constants; uppercase is a convention.

````python
name = "Hemil"
age = 22
height = 5.8
is_student = True

print(name, age, height, is_student)

x = 10
x = "Python"
print(x)
````

## 4. Objects, `type()` and `id()`

Everything in Python is an **object**.

- `type(value)` tells you the object's type.
- `is` checks object identity.
- `==` checks value equality.

Do not use `id()` as a business identifier; it is mainly useful for understanding/debugging object identity.

````python
x = 100

print(type(x))

a = [1, 2]
b = a
c = [1, 2]

print(a == b)  # same values
print(a == c)  # same values
print(a is b)  # same object
print(a is c)  # different objects
````

## 8. `None`

`None` represents the absence of a value.

It is **not** the same as `0`, `False`, or an empty string.

Prefer:

```python

    ...
```

instead of:

```python
if result == None:
    ...
```

`is None` is the conventional identity check for `None`.

````python
result = None

print(result)
print(type(result))

if result is None:
    print("No result yet")
````

## 12. Conditional Statements

### `if`



## 13. `match` Statement (Modern Alternative to Long Equality Checks)


Instead of many equality checks:

```python
if command == "start":
    ...
elif command == "stop":
    ...
```

we can sometimes use:

```python
match command:
    case "start":
        ...
    case "stop":
        ...
```

Use `match` when it makes multiple patterns clearer; do not replace every simple `if` with it.

````python
command = "start"

match command:
    case "start":
        print("Starting")
    case "stop":
        print("Stopping")
    case _:
        print("Unknown command")
````

## 14. `for` Loops

```python
range(stop)
```

## 15. `while` Loops

A `while` loop repeats while a condition is true.

```python

    print(i)
    i += 1
```

Be careful to update the condition/state, otherwise you can create an infinite loop.

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

## 16. `break`, `continue`, and `pass`

### `break`
Immediately exits the nearest loop.

### `continue`
Skips the current iteration and continues the loop.

### `pass`
Does nothing. It is a placeholder used where Python requires a statement.

```python
def future_function():
    pass
```

`pass` is not the same as `continue` or `break`.

````python
for i in range(10):
    if i == 5:
        break
    print(i)

print("---")

for i in range(5):
    if i == 2:
        continue
    print(i)

print("---")

def future_function():
    pass

print("pass example completed")
````

## 17. `else` with Loops

Python allows `else` on `for` and `while` loops.

The loop `else` runs when the loop finishes normally, **not** when it is stopped by `break`.

This is useful for search operations.

````python
numbers = [1, 3, 5, 7]
target = 4

for number in numbers:
    if number == target:
        print("Found")
        break
else:
    print("Not found")
````

## 19. String Methods

Common methods:

- `.lower()`
- `.upper()`
- `.title()`
- `.capitalize()`
- `.strip()`
- `.lstrip()`
- `.rstrip()`
- `.replace()`
- `.split()`
- `.join()`
- `.startswith()`
- `.endswith()`
- `.find()`
- `.count()`

### Which one to use?

- `strip()` → remove surrounding whitespace
- `replace()` → replace text
- `split()` → turn text into a list
- `join()` → combine strings
- `find()` → find a substring position
- `startswith()/endswith()` → test prefixes/suffixes
- `in` → simple substring membership test

For checking presence, prefer:

```python
if "Python" in text:
```

instead of using `find()` just to get a boolean.

````python
text = "  hello python world  "

print(text.strip())
print(text.upper())
print(text.title())
print(text.replace("python", "Python"))

words = text.strip().split()
print(words)

print("-".join(words))

print(text.strip().startswith("hello"))
print(text.strip().endswith("world"))
print(text.find("python"))
print(text.count("o"))
print("python" in text)
````

## 20. f-Strings

f-strings are the preferred readable way to insert values into strings.

```python
name = "Hemil"
age = 22
print(f"My name is {name} and I am {age}.")
```

You can also put expressions inside `{}`:

```python
print(f"Total: {10 + 20}")
```

Formatting:

```python
price = 1234.5678
print(f"{price:.2f}")
```

Alternatives:
- `str.format()` — older but still useful.
- `%` formatting — legacy style; generally avoid for new code unless working with existing code.

Prefer f-strings for modern Python code.

````python
name = "Hemil"
age = 22
price = 1234.5678

print(f"My name is {name} and I am {age}.")
print(f"Total: {10 + 20}")
print(f"Price: ₹{price:.2f}")

print("My name is {} and I am {}.".format(name, age))
````

## 21. Lists

A list is:
- ordered
- mutable
- allows duplicates
- supports indexing and slicing

Create:

```python
languages = ["Python", "JavaScript", "Java"]
```

Common operations:
- `append()` → add one item at end
- `extend()` → add multiple items
- `insert()` → add at an index
- `remove()` → remove first matching value
- `pop()` → remove and return by index (last item by default)
- `clear()` → remove all
- `sort()` → sort the list in place
- `reverse()` → reverse in place

Use a list when you need an ordered collection that may change.

````python
languages = ["Python", "JavaScript", "Java"]

languages.append("C++")
languages.extend(["Go", "Rust"])
languages.insert(1, "TypeScript")

print(languages)

languages.remove("Java")
last = languages.pop()

print("Removed:", last)
print(languages)

numbers = [5, 2, 9, 1]
numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
````

## 22. List Indexing and Slicing

```python
numbers = [10, 20, 30, 40, 50]
```

Access:
- `numbers[0]`
- `numbers[-1]`

Slice:
- `numbers[1:4]`
- `numbers[:3]`
- `numbers[2:]`
- `numbers[::2]`
- `numbers[::-1]`

### Copying matters

```python
b = a
```

does not create a new list; both names refer to the same list.

Use:

```python
b = a.copy()
```

or:

```python
b = a[:]
```

for a shallow copy.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])

a = [1, 2, 3]
b = a
c = a.copy()

a.append(4)

print("a:", a)
print("b:", b)  # changed because b references the same list
print("c:", c)  # independent shallow copy
````

## 23. List Comprehensions

A list comprehension is a compact way to create a list.

Normal loop:

```python
squares = []
for x in range(5):
    squares.append(x * x)
```

Comprehension:

```python
squares = [x * x for x in range(5)]
```

With condition:

```python
even = [x for x in range(10) if x % 2 == 0]
```

Use comprehensions when they remain readable. For complex multi-step logic, a normal loop is often clearer.

````python
squares = [x * x for x in range(5)]
even_numbers = [x for x in range(10) if x % 2 == 0]

print(squares)
print(even_numbers)
````

## 24. Tuples

A tuple is an ordered, immutable collection.

```python
point = (10, 20)
```

A one-item tuple needs a comma:

```python
single = (10,)
```

Tuples are useful for fixed groups of values and can be unpacked easily.

### List vs tuple

Use:
- `list` → collection changes
- `tuple` → fixed/immutable grouping

A tuple itself is immutable, but it can contain mutable objects.

````python
point = (10, 20)
single = (10,)

print(point[0])
print(single)

x, y = point
print(x, y)
````

## 25. Sets

A set stores unique elements.

```python
numbers = {1, 2, 3, 3}
```

The duplicate disappears.

Sets are useful for:
- removing duplicates
- fast membership checks
- mathematical set operations

Important operations:
- union: `|`
- intersection: `&`
- difference: `-`
- symmetric difference: `^`

Sets are unordered collections; do not rely on a specific display order.

````python
numbers = {1, 2, 3, 3, 4}
print(numbers)

a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric difference:", a ^ b)

print(2 in a)
````

## 26. Dictionaries

A dictionary stores key-value pairs.

```python
student = {
    "name": "Hemil",
    "age": 22,
    "course": "MSc IT"
}
```

Access:
```python
student["name"]
```

Safer optional lookup:
```python
student.get("email")
```

`get()` returns `None` (or a supplied default) if the key is missing, while `student["email"]` raises `KeyError`.

Useful methods:
- `.get()`
- `.keys()`
- `.values()`
- `.items()`
- `.update()`
- `.pop()`
- `.setdefault()`

Dictionaries are especially important for JSON and API data.

````python
student = {
    "name": "Hemil",
    "age": 22,
    "course": "MSc IT"
}

print(student["name"])
print(student.get("email"))
print(student.get("email", "Not provided"))

student["city"] = "Surat"
student["age"] = 23
student.update({"skills": ["Python", "MERN"]})

print(student)

for key, value in student.items():
    print(key, "=>", value)
````

## 27. Dictionary Access Alternatives

### Direct indexing

```python
value = data["name"]
```

Use when the key **must exist**.

### `.get()`

```python
value = data.get("name")
```

Use when the key may be missing.

### Membership check

```python
if "name" in data:
    print(data["name"])
```

Use when you want explicit control.

### `setdefault()`

```python
data.setdefault("skills", [])
```

It inserts the default only when the key is missing.

Choose based on whether a missing key is an error or an expected situation.

````python
data = {"name": "Hemil"}

print(data["name"])
print(data.get("email"))

if "name" in data:
    print("Name exists:", data["name"])

data.setdefault("skills", [])
data["skills"].append("Python")

print(data)
````

## 28. Unpacking

Python can unpack collections into variables.

### Sequence unpacking

```python

```

### Star unpacking

```python
first, *middle, last = [1, 2, 3, 4, 5]
```

### Dictionary unpacking

```python
person = {"name": "Hemil", "age": 22}

```

Unpacking is widely used in functions, loops, API data, and modern Python code.

````python
a, b, c = [1, 2, 3]
print(a, b, c)

first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)

person = {"name": "Hemil", "age": 22}
copy_person = {**person, "city": "Surat"}
print(copy_person)
````

## 29. Looping Through Collections

### List

```python
for item in items:
    ...
```

### Dictionary keys

```python
for key in data:
    ...
```

### Dictionary key/value pairs

```python
for key, value in data.items():
    ...
```

### `enumerate()`

Use when you need both index and value.

```python
for index, value in enumerate(items):
    ...
```

### `zip()`

Use to iterate over multiple sequences together.

```python
for name, score in zip(names, scores):
    ...
```

These are preferable to manually maintaining indexes in many situations.

````python
languages = ["Python", "JavaScript", "Java"]

for language in languages:
    print(language)

for index, language in enumerate(languages, start=1):
    print(index, language)

names = ["A", "B", "C"]
scores = [80, 90, 75]

for name, score in zip(names, scores):
    print(name, score)
````

## 31. Basic Exception Awareness

Some operations can fail.

Examples:
- `int("abc")` → `ValueError`
- `data["missing"]` → `KeyError`
- `10 / 0` → `ZeroDivisionError`
- `items[99]` → `IndexError`

For now, understand that errors are objects raised by Python. Later, when learning functions/projects, use `try/except` to handle expected failures.

Example:

```python
try:
    age = int(input("Age: "))
except ValueError:
    print("Please enter a valid integer.")
```

Avoid using broad `except:` unless you have a strong reason. Catch the expected exception type.

````python
def safe_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(safe_int("22"))
print(safe_int("abc"))
````

## 34. Scope Basics

Python has different name scopes. At the beginner level, remember:

- Local variable → inside a function
- Global variable → defined at module level
- Built-in names → provided by Python

Example:

```python
x = "global"

def demo():
    x = "local"
    print(x)

demo()
print(x)
```

The function's local `x` does not replace the global `x`.

When we study Functions, we'll cover scope, `global`, `nonlocal`, closures, and LEGB in detail.

````python
x = "global"

def demo():
    x = "local"
    print("Inside:", x)

demo()
print("Outside:", x)
````

### `sorted()` vs `.sort()`

```python
sorted_list = sorted(numbers)
```

returns a **new list**.

```python
numbers.sort()
```

changes the existing list and returns `None`.

Use `sorted()` when you need to preserve the original order.

````python
numbers = [5, 1, 4, 2]

print(len(numbers))
print(sum(numbers))
print(min(numbers))
print(max(numbers))
print(sorted(numbers))
print(numbers)

numbers.sort()
print(numbers)

print(any([False, False, True]))
print(all([True, True, True]))
````

## 34. Common Alternatives — What Should I Use?

| Need | Preferred option | Alternative / note |
|---|---|---|
| Insert a value into text | f-string | `.format()` for older code |
| Check if item exists | `in` | `.find()` when you need position |
| Safe dictionary lookup | `.get()` | `[]` when missing key should raise an error |
| Create transformed list | List comprehension | Normal `for` loop for complex logic |
| Sort without modifying original | `sorted()` | `.sort()` modifies the list |
| Add one list item | `append()` | `insert()` when position matters |
| Add many list items | `extend()` | `+` creates another list |
| Fixed collection | tuple | list if it must change |
| Unique values | set | list if duplicates/order matter |
| Key-value data | dict | list of pairs for special cases |
| Check no value | `is None` | avoid `== None` |
| Boolean condition | `if value:` | `if len(value) > 0` is often unnecessary |
| Exact decimal arithmetic | `Decimal` | float for ordinary numeric work |
| Iterate with index | `enumerate()` | manual counter only when needed |
| Iterate two sequences | `zip()` | index-based loop if synchronization logic requires it |

## 35.Student Grade Calculator

This combines:
- input
- type conversion
- variables
- conditions
- dictionary
- f-strings
- basic validation

The function used here is intentionally simple. Later, we'll rebuild this project properly using Functions and OOP.

````python
name = input("Enter student name: ").strip()
marks = float(input("Enter marks (0-100): "))

if not 0 <= marks <= 100:
    print("Marks must be between 0 and 100.")
else:
    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    else:
        grade = "D"

    student = {
        "name": name,
        "marks": marks,
        "grade": grade
    }

    print("\nStudent Details")
    print(f"Name: {student['name']}")
    print(f"Marks: {student['marks']}")
    print(f"Grade: {student['grade']}")
````

## 36. Simple Calculator

This version uses conditions. Later, Functions will make this much cleaner.

````python
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ").strip()
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 == 0:
        result = "Cannot divide by zero"
    else:
        result = num1 / num2
else:
    result = "Invalid operator"

print("Result:", result)
````
