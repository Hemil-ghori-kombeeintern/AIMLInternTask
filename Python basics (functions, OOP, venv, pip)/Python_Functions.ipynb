## 1. What is a Function?

A function is a reusable block of code that performs a specific task.

Instead of repeating:

```python
print("Hello Hemil")
print("Hello Hemil")
```

create it once:

```python
def greet():
    print("Hello Hemil")
```

and call it whenever needed:

```python
greet()
```

### Why use functions?
- Code reuse
- Avoid duplicate code
- Better organization
- Readability
- Easier testing
- Easier maintenance
- Break large programs into smaller tasks

````python
def greet():
    print("Hello Hemil")

greet()
greet()
greet()
````

## 2. Defining vs Calling

**Defining** creates the function:

```python
def greet():
    print("Hello")
```

**Calling** executes it:

```python
greet()
```

Flow:

```text
Define → Function object created → Call → Execute body → Optional return
```

````python
def greet():
    print("Hello")

print("Before call")
greet()
print("After call")
````

## 3. Basic Syntax

```python
def function_name():
    # function body
```

- `def` → defines a function
- `function_name` → function name
- `()` → parameters go here
- `:` → begins the block
- indentation → defines the body

````python
def welcome():
    print("Welcome to Python!")

welcome()
````

## 4. Parameters and Arguments

A **parameter** is a variable in the function definition.

An **argument** is the actual value passed during the call.

```python
def greet(name):       # name = parameter
    print(f"Hello {name}")

greet("Hemil")         # "Hemil" = argument
```

````python
def greet(name):
    print(f"Hello {name}")

greet("Hemil")
greet("Rahul")
````

## 5. Multiple Parameters

```python
def add(a, b):
    return a + b
```

You can have as many parameters as the function needs, but keep functions focused and readable.

````python
def add(a, b):
    return a + b

print(add(10, 20))

def student_info(name, age, course):
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Course: {course}")

student_info("Hemil", 22, "MSc IT")
````

## 6. `return`

`return` sends a result back to the caller and immediately exits the current function.

```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```


````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

## 7. `print()` vs `return`

`print()` displays a value.

`return` gives a value back to the caller.

```python
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b
```

Use `return` when another part of the program needs to use the result. Use `print()` for presentation/debugging.

````python
def add_print(a, b):
    print(a + b)

result = add_print(10, 20)
print("Returned:", result)

def add_return(a, b):
    return a + b

result = add_return(10, 20)
print("Returned:", result)
print("Reusable:", result * 2)
````

## 8. No Explicit Return

If a function does not explicitly return a value, Python returns `None`.

```python
def greet():
    print("Hello")

result = greet()
print(result)
```

````python
def greet():
    print("Hello")

result = greet()
print("Result:", result)
print("Type:", type(result))
````

## 9. Multiple Return Values

Python can return multiple values. They are packaged as a tuple.

```python
def calculate(a, b):
    return a + b, a - b
```

You can unpack them:

```python
addition, subtraction = calculate(10, 5)
```

````python
def calculate(a, b):
    return a + b, a - b

result = calculate(10, 5)
print(result)
print(type(result))

addition, subtraction = calculate(10, 5)
print(addition, subtraction)
````

## 10. Early Return

`return` can be used as a guard clause.

```python
def check_age(age):
    if age < 18:
        return "Minor"
    return "Adult"
```

Once a return executes, the function stops.

````python
def check_age(age):
    if age < 18:
        return "Minor"
    return "Adult"

print(check_age(15))
print(check_age(22))
````

## 11. Default Parameters

A parameter can have a default value.

```python
def greet(name="Guest"):
    print(f"Hello {name}")
```

If no argument is supplied, `"Guest"` is used.

### Rule
Required parameters must come before default parameters.

Correct:
```python
def student(name, age=18):
    ...
```

Incorrect:
```python
# def student(age=18, name):
#     ...
```

````python
def greet(name="Guest"):
    print(f"Hello {name}")

greet("Hemil")
greet()
````

## 12. Positional Arguments

Arguments can be matched by position.

```python
def student(name, age):
    print(name, age)

student("Hemil", 22)
```

The first argument goes to the first parameter, the second to the second.

````python
def student(name, age):
    print(f"Name: {name}, Age: {age}")

student("Hemil", 22)
````

## 13. Keyword Arguments

You can specify parameter names explicitly:

```python
student(name="Hemil", age=22)
```

Benefits:
- More readable
- Order can be changed
- Meaning of values is clear

````python
def student(name, age, course):
    print(name, age, course)

student(name="Hemil", age=22, course="MSc IT")
student(course="MSc IT", name="Hemil", age=22)
````

## 14. Positional + Keyword Arguments

Positional arguments must come before keyword arguments.

Valid:

```python
student("Hemil", age=22, course="MSc IT")
```

Invalid:

```python
# student(name="Hemil", 22, course="MSc IT")
```

````python
def student(name, age, course):
    print(name, age, course)

student("Hemil", age=22, course="MSc IT")
````

## 15. Keyword-Only Parameters

Use `*` to force later parameters to be keyword-only.

```python
def create_user(name, *, age, city):
    ...
```

Valid:

```python
create_user("Hemil", age=22, city="Surat")
```

This is useful when explicit named options improve readability.

````python
def create_user(name, *, age, city):
    return {"name": name, "age": age, "city": city}

print(create_user("Hemil", age=22, city="Surat"))
````

## 16. `*args`

`*args` collects extra positional arguments into a tuple.

```python
def add_numbers(*args):
    return sum(args)
```

The name `args` is conventional; `*numbers` would also be valid.

````python
def show_args(*args):
    print(args)
    print(type(args))

show_args(10, 20, 30)

def add_numbers(*numbers):
    return sum(numbers)

print(add_numbers(10, 20, 30, 40))
````

## 17. `**kwargs`

`**kwargs` collects extra keyword arguments into a dictionary.

```python
def student_info(**kwargs):
    print(kwargs)
```

Example:

```python
student_info(name="Hemil", age=22, city="Surat")
```

````python
def student_info(**kwargs):
    print(kwargs)
    print(type(kwargs))

student_info(name="Hemil", age=22, city="Surat")

def display_student(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

display_student(name="Hemil", age=22, course="MSc IT")
````

## 18. `*args` vs `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Collects | Positional arguments | Keyword arguments |
| Type | Tuple | Dictionary |
| Example | `func(1, 2, 3)` | `func(a=1, b=2)` |

Use them when the number of inputs is genuinely variable.

````python
def example(*args, **kwargs):
    print("args:", args)
    print("kwargs:", kwargs)

example(1, 2, 3, name="Hemil", city="Surat")
````

## 19. Combining Parameters

A function can combine normal parameters, defaults, `*args`, keyword-only parameters, and `**kwargs`.

Example:

```python
def example(a, b=10, *args, c=20, **kwargs):
    ...
```

Not every function should use all of these. Use only what makes the function interface clearer.

````python
def example(a, b=10, *args, c=20, **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("c:", c)
    print("kwargs:", kwargs)

example(1, 2, 3, 4, c=30, name="Hemil", city="Surat")
````

## 20. Type Hints

Type hints describe expected types.

```python
def add(a: int, b: int) -> int:
    return a + b
```

They help:
- readability
- IDEs
- documentation
- static type checkers

**Important:** Python generally does not enforce type hints automatically at runtime.

````python
def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20))

def greet(name: str, age: int = 18) -> str:
    return f"{name} is {age} years old"

print(greet("Hemil", 22))
````

## 22. Local Scope

Variables created inside a function are normally local.

```python
def greet():
    message = "Hello"
```

`message` cannot normally be accessed outside that function.

This helps keep implementation details isolated.

````python
def greet():
    message = "Hello from inside"
    print(message)

greet()

# This would raise NameError:
# print(message)
````

## 23. Global Scope

A variable defined outside a function is in global/module scope.

A function can read it:

```python
name = "Hemil"

def greet():
    print(name)
```

For maintainable applications, prefer passing data into functions rather than relying heavily on global state.

````python
name = "Hemil"

def greet():
    print(f"Hello {name}")

greet()
````

## 24. `global`

If you assign to a global variable inside a function, you need `global`.

```python
count = 0

def increment():
    global count
    count += 1
```

However, prefer passing state and returning a new value when practical.

````python
count = 0

def increment_global():
    global count
    count += 1

increment_global()
increment_global()
print("Global:", count)

def increment(count):
    return count + 1

count = 0
count = increment(count)
count = increment(count)
print("Preferred style:", count)
````

## 25. Nested Functions

Useful for:

- decorators
- keeping implementation details local

````python
def outer():
    def inner():
        print("Inside inner function")
    inner()

outer()
````

## 26. Lambda Functions

A lambda is a small anonymous function.

Syntax:

```python
lambda parameters: expression
```

Example:

```python
square = lambda x: x * x
```

Equivalent:

```python
def square(x):
    return x * x
```

### When to use
Use lambda for short, simple expressions, often as a callback/key function.

For complex logic, prefer `def`.

````python
square = lambda x: x * x

print(square(5))

def square_def(x):
    return x * x

print(square_def(5))
````

## 28. Function as an Argument

A function can accept another function.

```python
def square(x):
    return x * x

def execute(function, value):
    return function(value)

print(execute(square, 5))
```

This is a higher-order function pattern and is important for callbacks and decorators.

````python
def square(x):
    return x * x

def execute(function, value):
    return function(value)

print(execute(square, 5))


````

## 29. Recursion

Recursion means a function calls itself.

Every recursive function needs a base case.

Example:

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

````python
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

print(factorial_recursive(5))
````

## 30. How Arguments Are Passed

A useful Python mental model is:

> Python passes object references by assignment.

For immutable values, assigning a new value to a local parameter does not change the caller's variable.

For mutable objects, the function can mutate the same object.

This distinction is important in real projects.

````python
def change_number(x):
    x = 100
    print("Inside:", x)

number = 10
change_number(number)
print("Outside:", number)

def add_item(items):
    items.append("Python")

languages = ["JavaScript"]
add_item(languages)
print("Mutable list:", languages)
````

## 31. Mutable Default Argument Problem

Avoid:

```python
def add_item(item, items=[]):
    items.append(item)
    return items
```

The default list is created once and reused.

Preferred:

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

This is a common Python interview question.

````python
def add_bad(item, items=[]):
    items.append(item)
    return items

print("Bad pattern:")
print(add_bad("A"))
print(add_bad("B"))

def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print("Correct pattern:")
print(add_item("A"))
print(add_item("B"))
````

## 32. Function Overloading

Python does not support traditional overloads by defining multiple functions with the same name and different parameter lists.

The later definition replaces the earlier one.

### Alternatives
- Default parameters
- `*args`
- `**kwargs`
- `functools.singledispatch` for specific type-based designs

Example:

```python
def add(a, b, c=0):
    return a + b + c
```

````python
def add(a, b, c=0):
    return a + b + c

print(add(10, 20))
print(add(10, 20, 30))
````

````python
def calculate_total(marks):
    return sum(marks)

def calculate_average(marks):
    if not marks:
        raise ValueError("Marks list cannot be empty")
    return sum(marks) / len(marks)

def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    return "D"

marks = [80, 90, 75, 85]

total = calculate_total(marks)
average = calculate_average(marks)
grade = calculate_grade(average)

print(f"Total: {total}")
print(f"Average: {average}")
print(f"Grade: {grade}")
````
