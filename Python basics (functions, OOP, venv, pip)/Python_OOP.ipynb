## 1. What is OOP?

**OOP = Object-Oriented Programming.**

OOP organizes programs around **objects** containing:

- **Data** → attributes
- **Behavior** → methods

Example:

```text
Student
├── name
├── age
├── course
└── study()
```

The four commonly taught pillars are:
1. Encapsulation
2. Inheritance
3. Polymorphism
4. Abstraction

## 2. Class and Object

A **class** is a blueprint.

```python
class Student:
    pass
```

An **object** is an instance of that class.

```python
student1 = Student()
student2 = Student()
```

```text
             Student class
              /       \
             ↓         ↓
        student1    student2
          object      object
```

````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(type(student1))
print(student1 is student2)
````

## 3. Attributes

Attributes are data stored on an object.

````python
class Student:
    pass

student = Student()
student.name = "Hemil"
student.age = 22

print(student.name)
print(student.age)
````

## 4. Methods

A method is a function defined inside a class.

````python
class Student:
    def study(self):
        print("Student is studying")

student = Student()
student.study()
````

## 5. `self`

`self` refers to the current object.

```python
student.study()
```

is conceptually similar to:

```python
Student.study(student)
```

`self` is a convention and should be used for the instance parameter.

````python
class Student:
    def introduce(self):
        print("Current object:", self)

student = Student()
student.introduce()

# Conceptually:
Student.introduce(student)
````

## 6. `__init__`

`__init__` initializes an object after it is created.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Then:

```python
student = Student("Hemil", 22)
```

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Hemil", 22)

print(student.name)
print(student.age)
print(student.__dict__)
````

## 7. Instance Variables

Instance variables belong to individual objects.

```text
student1.name → Hemil
student2.name → Rahul
```

Each object can have different values.

````python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Hemil")
student2 = Student("Rahul")

print(student1.name)
print(student2.name)
````

## 8. Instance Methods

Instance methods operate on a particular object and receive `self` automatically.

````python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

account1 = BankAccount(1000)
account2 = BankAccount(500)

account1.deposit(200)

print(account1.balance)
print(account2.balance)
````

## 9. Class Variables

A class variable is stored on the class and can be shared by instances.

```python
class Student:
    school = "ABC College"
```

````python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

s1 = Student("Hemil")
s2 = Student("Rahul")

print(s1.school)
print(s2.school)
print(Student.school)
````

## 10. Instance vs Class Variable

| Instance variable | Class variable |
|---|---|
| Usually `self.name` | Defined in class body |
| Object-specific | Class-level |
| Can differ between objects | Common class-level value |
| Example: `self.name` | Example: `Student.school` |

Be careful with mutable class variables such as lists because instances may share the same object.

````python
class Student:
    school = "ABC College"

    def __init__(self, name):
        self.name = name

s1 = Student("Hemil")
s2 = Student("Rahul")

s1.name = "Hemil Ghori"

print(s1.name)
print(s2.name)
print(Student.school)
````

# Encapsulation

## 11. Encapsulation

Encapsulation means keeping related data and behavior together and controlling how internal state is accessed.

Python uses conventions and properties rather than strict private access rules like some languages.

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

account = BankAccount(1000)
account.deposit(500)

print(account.balance)
````

## 12. Public, Protected and Private Conventions

### Public
```python

```

### Protected convention
```python
self._balance
```

Single underscore means internal-use convention.

### Private/name-mangled
```python

```

Double underscore triggers name mangling. It is not absolute security.

````python
class Example:
    def __init__(self):
        self.public = "Public"
        self._protected = "Internal convention"
        self.__private = "Name mangled"

obj = Example()

print(obj.public)
print(obj._protected)
print(obj._Example__private)

class Student:
    def __init__(self):
        self.__rivate = "Heldfdlo"

obj = Student()

print(obj._Student__rivate)
````

## 13. Getter and Setter

Traditional style:

```python
get_age()
set_age()
```

Python often prefers `@property` for attribute-like access.

````python
class Student:
    def __init__(self, age):
        self._age = age

    def get_age(self):
        return self._age

    def set_age(self, age):
        if age < 0:
            raise ValueError("Age cannot be negative")
        self._age = age

student = Student(22)
print(student.get_age())

student.set_age(23)
print(student.get_age())
````

## 14. `@property`

A property allows controlled access using normal attribute syntax.

````python
class Student:
    def __init__(self, age):
        self.age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value

student = Student(22)

print(student.age)
student.age = 23
print(student.age)
````

# Inheritance

## 15. Inheritance

Inheritance allows a child class to reuse and extend a parent class.

```text
Animal
   ↑
   │
  Dog
```

````python
class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    pass

dog = Dog()
dog.speak()
````

## 16. Parent and Child Class

Terminology:
- Parent / Base / Superclass → `Animal`
- Child / Derived / Subclass → `Dog`

A child can:
- use inherited methods
- add new methods
- override inherited methods

## 17. Inheritance with `__init__` and `super()`

Use `super()` to call parent behavior.

````python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

dog = Dog("Tommy", "Labrador")

print(dog.name)
print(dog.breed)
````

## 18. Method Overriding

A child class can replace a parent's method implementation.

````python
class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    def speak(self):
        print("Bark")

class Cat(Animal):
    def speak(self):
        print("Meow")

Dog().speak()
Cat().speak()
````

## 19. Types of Inheritance

### Single
```text
A → B
```

### Multilevel
```text
A → B → C
```

### Multiple
```text
A ─┐
   ├→ C
B ─┘
```

### Hierarchical
```text
    A
   / \
  B   C
```

### Hybrid
Combination of inheritance patterns.

````python
class A:
    pass

class B(A):
    pass

class C(B):
    pass

print(issubclass(B, A))
print(issubclass(C, A))
````

## 20. Multiple Inheritance

Python allows a class to inherit from multiple classes.

````python
class Father:
    def father_skill(self):
        print("Driving")

class Mother:
    def mother_skill(self):
        print("Cooking")

class Child(Father, Mother):
    pass

child = Child()
child.father_skill()
child.mother_skill()
````

# Polymorphism

## 21. Polymorphism

Polymorphism means different objects can respond to the same operation/interface.

```python
for animal in [Dog(), Cat()]:
    animal.speak()
```

````python
class Dog:
    def speak(self):
        return "Bark"

class Cat:
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
````

## 22. Duck Typing

Python commonly follows the idea:

> If an object has the required behavior, it can be used.

Inheritance is not always required.

````python
class Dog:
    def speak(self):
        return "Bark"

class Robot:
    def speak(self):
        return "Beep"

def make_speak(obj):
    print(obj.speak())

make_speak(Dog())
make_speak(Robot())
````

# Abstraction

## 23. Abstraction

Abstraction exposes the important interface while hiding implementation details.

Python supports formal abstraction using the `abc` module.

````python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class UPI(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using UPI"

payment = UPI()
print(payment.pay(500))
````

## 24. Abstract Class

An abstract class can define methods that subclasses must implement.

```python
from abc import ABC, abstractmethod
```

A subclass remains abstract until all required abstract methods are implemented.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rectangle = Rectangle(10, 5)
print(rectangle.area())
````

## 25. Encapsulation vs Abstraction

**Encapsulation**
- Bundles data and behavior
- Controls access to internal state

**Abstraction**
- Defines a useful interface
- Hides unnecessary implementation details

Simple memory trick:

```text
Encapsulation → HOW data is managed
Abstraction   → WHAT interface is exposed
```

# Dataclasses

## 26. Dataclasses

For classes mainly used to store data, `dataclasses` can remove boilerplate.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
```

Python generates useful methods such as `__init__`, `__repr__`, and equality behavior.

````python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int

student = Student("Hemil", 22)

print(student)
print(student.name)
print(student.age)
````

## 27. Dataclass Default Values

````python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int = 18

print(Student("Hemil"))
print(Student("Rahul", 21))
````

# Practical Project Example

## 28. Bank Account — OOP Example

This combines:
- Class and object
- Constructor
- Instance attributes
- Methods
- Encapsulation convention
- Property
- Validation

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = 0
        self.deposit(balance)

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient balance")
        self._balance -= amount

account = BankAccount("Hemil", 5000)
print("Balance:", account.balance)
account.deposit(1000)
account.withdraw(1500)

print("Owner:", account.owner)
print("Balance:", account.balance)
````

## 29. Payment System — Inheritance + Polymorphism + Abstraction

````python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class UPI(Payment):
    def pay(self, amount):
        return f"UPI payment of ₹{amount}"

class Card(Payment):
    def pay(self, amount):
        return f"Card payment of ₹{amount}"

def checkout(payment_method, amount):
    print(payment_method.pay(amount))

checkout(UPI(), 1000)
checkout(Card(), 1500)
````
