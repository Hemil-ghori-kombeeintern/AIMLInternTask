```python
raw_df = df.copy()
clean_df = df.copy()
```

```python
missing_markers = ["N/A", "NA", "NaN", "null", "NULL", "", " "]

clean_df = clean_df.replace(
    missing_markers,
    np.nan
)

clean_df.isna().sum()
```

```python
df.isna().sum()
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

clean_df.columns
```

```python
print("Duplicate rows before:", clean_df.duplicated().sum())

clean_df = clean_df.drop_duplicates().copy()

print("Duplicate rows after:", clean_df.duplicated().sum())
```

```python
text_columns = clean_df.select_dtypes(include="object").columns
display(text_columns)
for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()

clean_df[["name", "city", "gender", "department"]].head(10)
```

```python
clean_df["city"] = clean_df["city"].str.lower().str.strip()

city_map = {
    "surat": "Surat",
    "ahmedabad": "Ahmedabad",
    "mumbai": "Mumbai",
    "bombay": "Mumbai",
    "pune": "Pune",
    "delhi": "Delhi"
}

clean_df["city"] = clean_df["city"].replace(city_map)
```

```python
clean_df[[ "city" ]]
```

```python
clean_df["gender"] = clean_df["gender"].str.lower().str.strip()

gender_map = {
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
}


clean_df["gender"] = clean_df["gender"].replace(gender_map)
```

```python
clean_df[[ "gender" ]]
```

```python
clean_df["department"] = (
    clean_df["department"]
    .str.strip()
    .str.lower()
    .str.upper()
)
```

```python
clean_df[[ "department" ]]
```

```python
display(clean_df[["city", "gender", "department"]].drop_duplicates().sort_values(
    ["city", "gender", "department"],
    na_position="last"
))
```

```python
clean_df["name"] = clean_df["name"].replace(
    ["N/A", "NA", ""],
    np.nan
)

clean_df["name"] = clean_df["name"].str.strip().str.title()

clean_df["name"].head(10)
```

```python
clean_df["age"] = pd.to_numeric(
    clean_df["age"],
    errors="coerce"
)

clean_df["age"].dtype
```

```python
invalid_age = ~clean_df["age"].between(0, 100)

print("Invalid ages:")
display(clean_df.loc[invalid_age, ["id", "name", "age"]])

clean_df.loc[invalid_age, "age"] = np.nan
```

```python
clean_df["salary"] = (
    clean_df["salary"]
    .astype("string")
    .str.strip()
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
)

clean_df["salary"] = pd.to_numeric(
    clean_df["salary"],
    errors="coerce"
)

clean_df["salary"].dtype
```

```python
invalid_salary = clean_df["salary"] < 0

print("Invalid salary records:")
display(clean_df.loc[invalid_salary, ["id", "name", "salary"]])

clean_df.loc[invalid_salary, "salary"] = np.nan
```

```python
clean_df["joining_date"] = pd.to_datetime(
    clean_df["joining_date"],
    errors="coerce",
    dayfirst=True
)

clean_df["joining_date"].dtype
```

```python
clean_df["joining_year"] = clean_df["joining_date"].dt.year
clean_df["joining_month"] = clean_df["joining_date"].dt.month
clean_df["joining_day"] = clean_df["joining_date"].dt.day
clean_df["joining_day_name"] = clean_df["joining_date"].dt.day_name()

clean_df[[
    "joining_date",
    "joining_year",
    "joining_month",
    "joining_day",
    "joining_day_name"
]].head()
```

```python
clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())
clean_df["salary"] = clean_df["salary"].fillna(clean_df["salary"].median())

for col in ["name", "city", "gender", "department"]:
    clean_df[col] = clean_df[col].fillna("Unknown")

print(clean_df.isna().sum())
```

```python
for col in ["city", "gender", "department"]:
    print(f"\n{col}:")
    print(clean_df[col].value_counts(dropna=False))
```

```python
surat_records = clean_df[
    clean_df["city"] == "Surat"
]

high_salary = clean_df[
    clean_df["salary"] > 50000
]

ai_ml_adults = clean_df[
    (clean_df["department"] == "AI/ML") &
    (clean_df["age"] >= 18)
]

display(surat_records)
display(high_salary.head())
display(ai_ml_adults.head())
```

```python
clean_df["monthly_salary"] = clean_df["salary"] / 12

clean_df["age_group"] = np.where(
    clean_df["age"] >= 25,
    "25+",
    "Under 25"
)

clean_df[[
    "name",
    "age",
    "age_group",
    "salary",
    "monthly_salary"
]].head()
```

```python
Q1 = clean_df["salary"].quantile(0.25)
Q3 = clean_df["salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = clean_df[
    (clean_df["salary"] < lower_bound) |
    (clean_df["salary"] > upper_bound)
]

display(outliers[["id", "name", "salary"]])
```

```python
clean_df["salary"] = clean_df["salary"].clip(
    lower=lower_bound,
    upper=upper_bound
)

clean_df["salary"].describe()
```

```python
encoded_df = pd.get_dummies(
    clean_df,
    columns=["city", "gender", "department"],
    dtype=int
)

encoded_df.head()
```

```python
salary_min = clean_df["salary"].min()
salary_max = clean_df["salary"].max()

if salary_max != salary_min:
    clean_df["salary_scaled"] = (
        (clean_df["salary"] - salary_min) /
        (salary_max - salary_min)
    )
else:
    clean_df["salary_scaled"] = 0

clean_df[["salary", "salary_scaled"]].head()
```

```python
department_summary = clean_df.groupby("department"
    )["salary"].agg(
    ["count", "mean", "min", "max"]
)

department_summary
```

```python
print("Final shape:", clean_df.shape)
print("\nDuplicate rows:", clean_df.duplicated().sum())

print("\nMissing values:")
print(clean_df.isna().sum())

print("\nData types:")
print(clean_df.dtypes)

print("\nFinal statistics:")
display(clean_df.describe(include="all"))

```
