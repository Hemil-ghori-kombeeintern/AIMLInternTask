```python
text = """
   HELLO!!! I am learning NLP.
   Tokenization is IMPORTANT for AI models.
   I love learning machine learning 😊
   https://chatgpt.com/gsdc/g-p-6a959d75903c8191af0f1a03b9396b70-aimlinternship/c/6a8d5a1a-8a24-83e8-b300-932fa5462f81
"""

print(text)
```

```python
import re

def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Replace URLs with a placeholder
    text = re.sub(r"https?://\S+|www\.\S+", " [URL] ", text)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

cleaned_text = clean_text(text)
print(cleaned_text)
```

```python
import re

def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Replace URLs with a placeholder
    text = re.sub(r"https?://\S+|www\.\S+", " [URL] ", text)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

cleaned_text = clean_text(text)
print(cleaned_text)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab", quiet=True)

tokens = word_tokenize(cleaned_text)

print(tokens)
```

```python
from nltk.corpus import stopwords

nltk.download("stopwords", quiet=True)

stop_words = set(stopwords.words("english"))

tokens_without_stopwords = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print(tokens_without_stopwords)
```

```python
from nltk.tokenize import sent_tokenize

nltk.download("punkt_tab", quiet=True)

tokens = sent_tokenize(cleaned_text)

print(tokens)
```

```python
from nltk.corpus import stopwords

nltk.download("stopwords", quiet=True)

stop_words = set(stopwords.words("english"))

tokens_without_stopwords = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print(tokens_without_stopwords)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays", "studies","unbelievable"]

stemmed_words = [
    stemmer.stem(word) for word in words
]

print(stemmed_words)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("cars"))
print(lemmatizer.lemmatize("running", pos="v"))
print(lemmatizer.lemmatize("better", pos="a"))
```
