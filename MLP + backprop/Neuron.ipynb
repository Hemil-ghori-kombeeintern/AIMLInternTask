```python
import torch
import torch.nn as nn
```

```python
# input
torch.manual_seed(42)
inputs = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
print(neuron.weight)
print(neuron.bias)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
# verify ... formula
neuron.weight @ inputs + neuron.bias
```

```python
import torch
import torch.nn as nn
```

```python
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)
```

```python
print(model[0].weight)
print(model[0].bias)
```

```python
torch.manual_seed(42)
x = torch.tensor([
    [2.0, 3.0]
])
```

```python
output = model(x)
print(output)
```
