```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

documents = [
    "Employees can work from home three days per week.",
    "Employees receive their salary on the last working day.",
    "The company holiday calendar is available on the HR portal.",
    "Employees must use VPN when accessing company systems remotely.",
    "The company provides health insurance to full-time employees."
]

query = "What is the company's work from home policy?"

doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])

scores = cosine_similarity(
    query_embedding,
    doc_embeddings
)[0]

results = sorted(
    zip(documents, scores),
    key=lambda x: x[1],
    reverse=True
)

for rank, (doc, score) in enumerate(results, 1):
    print(f"{rank}. {score:.4f} - {doc}")
```

```python
models = {
    "MiniLM": "sentence-transformers/all-MiniLM-L6-v2",
    "MPNet": "sentence-transformers/all-mpnet-base-v2"
}

for name, model_name in models.items():

    model = SentenceTransformer(model_name)

    doc_embeddings = model.encode(documents)
    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        doc_embeddings
    )[0]

    ranking = scores.argsort()[::-1]

    print("\n", name)

    for rank, idx in enumerate(ranking, 1):
        print(
            rank,
            round(scores[idx], 4),
            documents[idx]
        )
```
